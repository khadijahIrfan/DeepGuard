```python
from PIL import Image
from PIL.ExifTags import TAGS

def extract_metadata(image_path):

    image = Image.open(image_path)

    exif_data = image._getexif()

    if exif_data is None:
        return {
            "status": "No Metadata Found",
            "reason": "Metadata may have been removed by WhatsApp, social media platforms, or image editing software."
        }

    metadata = {}

    for tag_id, value in exif_data.items():

        tag = TAGS.get(tag_id, tag_id)

        metadata[tag] = value

    return metadata
```

```python
result = extract_metadata(
    "WhatsApp Image 2026-06-23 at 3.25.32 PM.jpeg"
)

print(result)
```

```python
from PIL import Image
from PIL.ExifTags import TAGS

def extract_important_metadata(image_path):

    image = Image.open(image_path)

    exif = image._getexif()

    if not exif:
        return {"Status": "No Metadata Found"}

    metadata = {}

    for tag_id, value in exif.items():

        tag = TAGS.get(tag_id, tag_id)

        metadata[tag] = value

    important = {
        "Camera Brand": metadata.get("Make"),
        "Camera Model": metadata.get("Model"),
        "Date Taken": metadata.get("DateTime"),
        "Image Width": metadata.get("ExifImageWidth"),
        "Image Height": metadata.get("ExifImageHeight"),
        "Software": metadata.get("Software"),
        "GPS Info": metadata.get("GPSInfo")
    }

    return important
```

```python
result = extract_important_metadata(".trashed-1782993253-IMG_20260601_223809[1].jpg")

for key, value in result.items():
    print(key, ":", value)
```
